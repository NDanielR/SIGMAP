## **Carga de Data**

##### **Primero realizamos la carga de la informacion en el df**

```python
import pandas as pd
from IPython.display import display

excelPath = "Data\\Informacion sin limpiar.xlsx"

df_sin_limpiar = pd.read_excel(
    excelPath,
    sheet_name = "Análisis de ventas",
    header = 3,
    index_col = 0)

df_sin_limpiar = df_sin_limpiar.reset_index()# Eliminar la columna de índice

display(df_sin_limpiar.head())
```

##### **Separamos las columnas de productos para hacer la clasificacion**

```python
df_productos = df_sin_limpiar.iloc[:, 0:5]

display(df_productos.head())
```

##### **Cambiamos a mayuscula el encabezado**

```python
df_productos.columns = df_productos.columns.str.upper()# Convertir los nombres del emcabezado de las columnas a mayúsculas

display(df_productos.head())
```

##### **Cambiar los valores de las columnas a mayúsculas**

```python
# Convertir los valores de las columnas a minúsculas
df_productos["PRODUCTO"] = df_productos["PRODUCTO"].str.upper()
df_productos["CLASIFICACION I"] = df_productos["CLASIFICACION I"].str.upper()
df_productos["CLASIFICACION II"] = df_productos["CLASIFICACION II"].str.upper()
df_productos["CLASIFICACION III"] = df_productos["CLASIFICACION III"].str.upper()
df_productos["CLASIFICACION IV"] = df_productos["CLASIFICACION IV"].str.upper()

display(df_productos.head())
```

##### **Separamos los codigos de los productos**

```python
df_productos["CODIGO_PRODUCTO"] = (
    df_productos["PRODUCTO"]
    .str.extract(r"\[(.*?)\]", expand=False)
    .astype("string")
    .str.strip()
)

df_productos["NOMBRE_PRODUCTO"] = (
    df_productos["PRODUCTO"]
    .str.replace(r"\[.*?\]\s*", "", regex=True)
    .str.strip()
)

# Eliminar la columna original que tenía código + nombre juntos
df_productos = df_productos.drop(columns=["PRODUCTO"])

# Reordenar columnas
df_productos = df_productos[
    [
        "CODIGO_PRODUCTO",
        "NOMBRE_PRODUCTO",
        "CLASIFICACION I",
        "CLASIFICACION II",
        "CLASIFICACION III",
        "CLASIFICACION IV"
    ]
]

display(df_productos.head())
```

##### **Limpiamos los espacios de inio y fin y entre palabras del texto**

```python
columnas_limpiar = [
    "NOMBRE_PRODUCTO",
    "CLASIFICACION I",
    "CLASIFICACION II",
    "CLASIFICACION III",
    "CLASIFICACION IV"
]

df_productos[columnas_limpiar] = df_productos[columnas_limpiar].apply(
    lambda col: col.str.strip().str.replace(r"\s+", " ", regex=True)
)

df_productos["CODIGO_PRODUCTO"] = (
    df_productos["CODIGO_PRODUCTO"]
    .astype("string")
    .str.replace(r"\s+", "", regex=True)
)

display(df_productos.head())
```

```python
# Crear tabla maestra de productos
df_productos_maestro = df_productos.copy()

df_productos_maestro = df_productos_maestro[
    [
        "CODIGO_PRODUCTO",
        "NOMBRE_PRODUCTO",
        "CLASIFICACION I",
        "CLASIFICACION II",
        "CLASIFICACION III",
        "CLASIFICACION IV"
    ]
]

display(df_productos_maestro.head())
```

##### **Verificar los datos null del df**

```python
df_productos_maestro.isnull().sum()
```

##### **Cambiar los valores Null por NO APLICA**

```python
columnas_rellenar = [
    "CLASIFICACION II",
    "CLASIFICACION III",
    "CLASIFICACION IV"
]

# Convertir espacios vacíos o textos vacíos en valores nulos reales
df_productos_maestro[columnas_rellenar] = (
    df_productos_maestro[columnas_rellenar]
    .replace(r"^\s*$", pd.NA, regex=True)
)

# Ahora sí llenar con NO APLICA
df_productos_maestro[columnas_rellenar] = (
    df_productos_maestro[columnas_rellenar]
    .fillna("NO APLICA")
)
```
